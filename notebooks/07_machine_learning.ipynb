```python
import sys
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Add project root to path
project_root = Path.cwd().parent if 'notebooks' in str(Path.cwd()) else Path.cwd()
sys.path.insert(0, str(project_root))
sys.path.insert(0, str(project_root / 'src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from config import PROCESSED_DATA_DIR, MODELS_DIR, FIGURES_DIR
from ml_models import (
    DisasterFrequencyPredictor,
    ImpactSeverityPredictor,
    RiskScorePredictor,
    train_all_models
)

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('Set2')

print("✓ Libraries loaded")
```

## 1. Load Data

```python
# Load processed disaster data
data_file = PROCESSED_DATA_DIR / 'disaster_events_sendai.csv'

if data_file.exists():
    df = pd.read_csv(data_file)
    print(f"✓ Loaded {len(df)} disaster events")
    print(f"✓ Time period: {df['year'].min()} - {df['year'].max()}")
    print(f"✓ Hazard types: {df['hazardtype'].nunique()}")
    if 'level2' in df.columns:
        print(f"✓ Locations: {df['level2'].nunique()}")
else:
    print("⚠ Processed data not found. Please run 05_risk_modeling.ipynb first.")
    df = None
```

## 2. Train All Models

This will train:
- Frequency predictor (when disasters occur)
- Impact predictors (deaths, affected, economic loss)
- Risk score predictor (location-based risk)

```python
if df is not None:
    print("="*80)
    print("TRAINING MACHINE LEARNING MODELS")
    print("="*80)
    print()
    
    # Train all models
    models = train_all_models(df, save_models=True)
    
    print("\n" + "="*80)
    print("MODEL PERFORMANCE SUMMARY")
    print("="*80)
    
    # Display metrics
    for model_name, model_data in models.items():
        print(f"\n{model_name.upper().replace('_', ' ')}:")
        metrics = model_data['metrics']
        for metric_name, value in metrics.items():
            if metric_name != 'feature_importance' and isinstance(value, (int, float)):
                print(f"  {metric_name}: {value:.4f}")
```
